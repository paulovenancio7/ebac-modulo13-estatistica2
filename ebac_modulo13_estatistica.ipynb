```python
import matplotlib.pyplot as plt
import pandas as pd
import plotly.express as px
import seaborn as sns

# Install kaleido for Plotly image export
!pip install -U kaleido

# 1. Carregar a base de dados
df = pd.read_csv('/content/MODULO7_PROJETOFINAL_BASE_SUPERMERCADO - MODULO7_PROJETOFINAL_BASE_SUPERMERCADO (1).csv (1).csv', delimiter=',')

# ---------------------------------------------------------
# CÁLCULOS (Exercícios 1 e 2)
# ---------------------------------------------------------
estatisticas = (
    df.groupby('Categoria')['Preco_Normal']
    .agg(['mean', 'median', 'std'])
    .round(2)
)
print('=== RESULTADOS DE MÉDIA, MEDIANA E DESVIO PADRÃO ===')
print(estatisticas)

# ---------------------------------------------------------
# GRÁFICO 1: Boxplot da categoria 'lacteos' (Maior Desvio Padrão)
# ---------------------------------------------------------
plt.figure(figsize=(10, 5))
sns.boxplot(x=df[df['Categoria'] == 'lacteos']['Preco_Normal'], color='skyblue')
plt.title('Distribuição de Preço Normal - Categoria: lacteos')
plt.xlabel('Preço Normal')
plt.tight_layout()
plt.savefig('grafico1_boxplot_lacteos.png')  # Salva o arquivo da Imagem 1
plt.close()

# ---------------------------------------------------------
# GRÁFICO 2: Barras - Média de Desconto por Categoria
# ---------------------------------------------------------
media_desconto = (
    df.groupby('Categoria')['Desconto']
    .mean()
    .reset_index()
    .sort_values(by='Desconto', ascending=False)
)

plt.figure(figsize=(12, 6))
sns.barplot(data=media_desconto, x='Categoria', y='Desconto', palette='viridis', hue='Categoria', legend=False)
plt.title('Média de Desconto por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Média de Desconto')
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig('grafico2_barras_desconto.png')  # Salva o arquivo da Imagem 2
plt.close()

# ---------------------------------------------------------
# GRÁFICO 3: Mapa Interativo / Treemap
# ---------------------------------------------------------
df_mapa = (
    df.groupby(['Categoria', 'Marca'])['Desconto']
    .mean()
    .reset_index()
    .query('Desconto > 0')
)

fig = px.treemap(
    df_mapa,
    path=['Categoria', 'Marca'],
    values='Desconto',
    color='Desconto',
    color_continuous_scale='Blues',
    title='Média de Desconto por Categoria e Marca',
)
fig.write_html('grafico3_mapa_interativo.html')  # Salva o arquivo da Imagem 3
```
